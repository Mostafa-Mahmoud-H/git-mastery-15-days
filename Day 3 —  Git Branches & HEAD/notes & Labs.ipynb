### ***🌿 Git Branches & HEAD***
---
1. Branch
* A branch in Git is a **movable reference** to a **commit** that provides an **independent line of development** within a repository.
---
* ***`git branch`***
* is a Git command used to *create,* *list*, *rename*, and *delete* branch references.
    1. Create Branch : 
        * **`git branch <branch-name>`**
    2. List branches:
        * **`git branch`**  
           * The `*` indicates the current branch (* main)
        * **`git branch -v`** for more details
    3. Delete Branch:
        * **`git branch -d <branch name>`** secure  
        * **`git branch -D <brance name>`** Force delete
---
* History example: A___B____C <- main branch <- HEAD 
1. **`git branch new`** 
    * A___B____C <- main branch  <- HEAD  
    * A___B____C <- new branch 
2. **`git switch new`** now head points to new branch rather than main .  
    * changes the current branch and updates the working tree to match the selected branch.    
    * A___B____C <- main branch  
    * A___B____C <- new branch  <- HEAD  
3. Create new branch +  Switch in one line : 
    * **`git switch -c <branch_name>`**
---
* **What Happens When We Commit?**

* Suppose: `A → B → C <- features` 

* We make a new commit on `feature`:
* The history becomes: `A → B → C → D` main points to `C` and feature points to `D`
* The branch moved from C to D.
* The branch did not create a copy of the entire project.
---
* **Branch Divergence**
    * **Branch divergence** is the state in which **two branches** that shared a **common ancestor** develop different commits, causing their histories to split into separate lines of development.

    * Two branches start from the same commit, **then each branch creates different commits.**
* **Common Ancestor** :
    * **A common ancestor is a commit that is reachable from the histories of two or more branches.**
