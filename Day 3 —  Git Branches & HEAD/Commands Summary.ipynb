# 🔹 Commands Reference — Day 3

1. `git branch` : This command is used to list the branches in the repository. It shows all local branches and marks the currently checked-out branch with an asterisk (`*`). It can also be used to create, rename, and delete branches depending on the provided options.

2. `git branch <branch-name>` : This command is used to create a new branch. The new branch is created as a reference pointing to the current commit, but it does not switch the working directory to the new branch.

3. `git switch <branch-name>` : This command is used to switch from the current branch to another existing branch. It updates `HEAD` and the working directory to reflect the selected branch.

4. `git switch -c <branch-name>` : This command is used to create a new branch from the current commit and immediately switch to it.

5. `git branch -v` : This command is used to display local branches along with the commit each branch currently points to. It shows the branch name, abbreviated commit hash, and the latest commit message.

6. `git branch -d <branch-name>` : This command is used to safely delete a local branch. Git normally prevents deletion if the branch contains commits that have not been merged.

7. `git branch -D <branch-name>` : This command is used to forcefully delete a local branch, even if it contains commits that have not been merged. It should be used carefully.
