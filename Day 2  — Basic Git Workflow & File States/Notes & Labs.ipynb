### ***Here’s an important question:***

* How does **`Git`** actually know that a **file has changed?** What are the different states a file can be in? And how can I tell exactly what changed?
---
1. ***File State*** 
* A file state in Git describes the **relationship** between the current version of a file in the *`working tree`*, the version staged in the *`index`*, and the version recorded in the *`repository`* or *`HEAD`*.

    1. **Untracked** : is a file in the working tree that is **not currently represented in the repository's tracked file set**.
        * **`Working Tree`** -> **New File Created**, **`Index`** -> **Not Staged**, **`Repository(HEAD)`** -> **Not Committed**

    2. **Tracked** : is a file that **Git is aware of** and is being monitored for changes.
        * **`Working Tree`** -> **New Version**, **`Index`** -> **New Version**, **`Repository(HEAD)`** -> **New Version**
        * Tracked files can be in **one of three states**:
            - **Modified** : The file has been changed in the working directory but has not yet been staged for commit.
                * **`Working Tree`** -> **New Version**, **`Index`** -> **Old Version**, **`Repository(HEAD)`** -> **Old Version**
            - **Staged** : The file has been modified and the changes have been added to the staging area (index) in preparation for a commit.
                * **`Working Tree`** -> **New Version**, **`Index`** -> **New Version**, **`Repository(HEAD)`** -> **Old Version**
            - **Committed** : The file's current state has been saved in the repository's history (HEAD).
                * **`Working Tree`** -> **New Version**, **`Index`** -> **New Version**, **`Repository(HEAD)`** -> **New Version**
                
    3. **Clean Working Tree** : A clean working tree means that there are no changes in the working directory compared to the last commit or no unstaged changes in the working directory. All tracked files are in their committed state, and there are no untracked files present.    
---


### **`git diff`**
* `git diff` displays **differences between two states of files** by showing **line-level changes**.
*  What does `git diff` compare? **Working Tree and Index**. It shows the changes that have been made in the working directory but have not yet been staged for commit.(Unstaged Changes)

### **`git diff --staged`**
* `git diff --staged` (or `git diff --cached`) shows the differences between the **staged changes in the index** and the **last committed state in the repository (HEAD)**. It displays what will be included in the next commit.(Staged Changes)
---
### **Commit Message** 
* A commit message is a **human-readable description** associated with a commit that explains the purpose or effect of the recorded changes, A commit should represent one logical change whenever practical.
---
#### **Atomic Commit**
* An atomic commit is a commit that represents a **single, self-contained change** to the codebase. It should be focused on one specific task or feature, making it easier to understand, review, and revert if necessary. Atomic commits help maintain a clean and organized project history.

### **git log**    
* `git log` shows the commit history of the repository, displaying a list of commits in reverse chronological order. Each commit entry includes information such as the commit hash, author, date, and commit message. It allows developers to review the evolution of the codebase and understand the changes made over time.
* `git log --oneline` shows a compact view of the commit history with each commit on a single line.
