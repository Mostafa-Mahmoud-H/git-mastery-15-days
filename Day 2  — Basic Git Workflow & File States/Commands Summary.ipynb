1. `git status` : This command is used to display the state of the working directory and the staging area. It shows which changes have been staged, which haven't, and which files aren't being tracked by Git.
2. `git diff` : This command is used to show the differences between the working directory and the staging area. It displays the changes that have been made to files that are tracked by Git but have not yet been staged for commit.
3. `git diff --staged` : This command is used to show the differences between the staging area and the last commit. It displays the changes that have been staged for commit but have not yet been committed to the repository.
4. `git log` : This command is used to show the commit history of the repository. It displays a list of commits, along with their commit messages, author information, and timestamps.
5. `git log --oneline` : This command is used to show a condensed version of the commit history. It displays a list of commits in a single line format, showing only the commit hash and the commit message.
