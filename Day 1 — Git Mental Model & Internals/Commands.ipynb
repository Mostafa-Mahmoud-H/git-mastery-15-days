1. `git init` : Initializes a new Git repository in the current directory. This command creates a `.git` directory that contains all the necessary files and data for version control.
2. `git add <file>` : Adds changes made to a specific file to the staging area, preparing it for the next commit. You can also use `git add .` to stage all changes in the current directory.
3. `git status` : Displays the current status of the working directory and staging area, showing which files have been modified, added, or deleted.
4. `git commit -m "<message>"` : Creates a new commit with the changes that have been staged, along with a descriptive message explaining the changes made.
5. `git log` : Shows the commit history of the repository, including commit hashes, authors, dates, and commit messages.
