#### ***🧠 Git Mental Model & Internals***
---
#### **What is Git?**
* Git is a **distributed version control system** that **tracks changes** in files and **coordinates work** on those files **among multiple people**.
* **You can find out:**
    * What the project looked like before?
    * What changed?
    * Who made the change?
    * When it happened?
    * Can you revert to a previous state?
    * Can you compare two versions?
    * Can you work with other people?
* **`Git`** runs primarily on your local machine and does not require a network connection to function. It is designed to handle everything from small to very large projects with speed and efficiency.
* **`GitHub`** : is a **cloud-based platform** for **hosting Git repositories** and **enabling collaboration around them**.  
    * **collaboration means** that multiple people can work on the same project at the same time, and **`GitHub`** provides tools to manage that collaboration effectively like `pull requests`, `code reviews`, `Actions`, and `issue tracking`.

---
* **Git Components:**
1. **Git Repository:** is a **data structure** that **stores the complete history** and **metadata of a project** under Git version control.
    * We use command `git init` to create a new Git repository.
    * What `git init` does is it creates a **hidden directory** called `.git` in the root of your project. This directory contains all the information about the repository, including the **commit history**, **branches**, and **configuration settings**.

***
* *****Lab-1*****:
    * Let's create our first repository.
        * In PowerShell:
            * `mkdir git-day-01`
            * `cd git-day-01`
            * `git init`
* after running `git init`, you will see a message indicating that an **empty Git repository has been initialized in the `.git` directory**. You can running `ls -Force` to see the hidden `.git` directory.
* The `.git` directory is the heart of the repository, Inside it, **Git stores**: `Objects`, `References`, `HEAD`, `Configuration`, `The Index`, and more.

---
2. **Working Tree :** is the **set of files** and **directories** currently checked out in a Git repository and available for you to edit.
* in a simple words, the working tree is the **current state of your project** that you see in your file system. It represents the files and directories that you are actively working on, and it reflects the latest commit in the repository.
---
3. **Staging Area (Index):** is an **intermediate area** where you can **prepare changes** before committing them to the Git repository.
* The staging area allows you to **selectively choose** which changes you want to include in your next commit.
* For example, if you have made changes to multiple files, you can add only specific files to the staging area, and then commit those changes while leaving the other changes uncommitted.
* You can add files to the staging area using the `git add` command.
---
4. **Commit:** is a **snapshot of the changes** in the staging area that you want to save to the Git repository.
* Each commit represents a **specific point in the history of the project** and contains ***information*** about the **changes made**, **the author of the changes**, and a **unique identifier** (**commit hash**).
* You can create a commit using the `git commit` command, which takes the changes from the staging area and saves them to the repository.
---
* *****Lab-2*****:
    * Let's create a file and commit it to the repository.
        * In PowerShell:
            * `echo "# Git Day 1" > README.md`
            * `git status`
                * You will see that the `README.md` file is **`untracked`**.
                * **`untracked`** files are files that are not yet being tracked by Git. They are new files that have been created in the working directory but have not been added to the staging area or committed to the repository.
            * `git add README.md`    
            * `git status`
                * You will see that the `README.md` file is now **`staged`**.
                * **`staged`** files are files that have been added to the staging area and are ready to be committed to the Git repository. When you stage a file, you are telling Git that you want to include the changes made to that file in your next commit.
            * `git commit -m "docs: add initial README"`    
                * You will see a message indicating that a new commit has been created with the specified commit message. The commit will have a unique identifier (commit hash) and will be added to the Git repository's history.
            * `git log`   
                * You will see the commit history, including the commit you just made. The log will display the commit hash, author, date, and commit message for each commit in the repository's history.
---
* **Commit Hash** : is a **unique identifier** that Git generates for each commit. It is a **40-character string** composed of hexadecimal characters (0-9 and a-f) that represents the **SHA-1 hash** of the commit's contents and metadata.             
    * It allows you to **refer to a specific commit** in the repository's history and is used for various Git operations, such as checking out a specific commit, creating branches, and merging changes.
    * why is it important? Because it allows you to **track changes**, **revert to previous versions**, and **collaborate with others** effectively. It provides a way to uniquely identify and reference specific points in the project's history, making it easier to manage and understand the evolution of the codebase.
---
* File Life Cycle in Git:
    * File Created -> File is created in the working tree.
    * File Modified -> File is modified in the working tree.
    * File Staged -> File is added to the staging area using `git add`.
    * File Committed -> File is committed to the Git repository using `git commit`.
    * File Edited -> File is modified again in the working tree, the status of the file will change to **`modified`**. 
---
* **HEAD** :  is a **symbolic reference** that points to the **current branch** and the **latest commit** in that branch. It represents the **current state of the repository** and indicates which commit is currently checked out in the working tree.  
    * HEAD importance appears **when you have more then one branch in your repository**. It allows you to **switch between branches and keep track of the current branch you are working on**. When you create a new commit, HEAD moves to point to the new commit, indicating that it is now the latest commit in the current branch. 
    * **HEAD Refers to the latest commit on the current branch**, and it is used by Git to determine the base for new commits and to manage the state of the working tree and staging area. 
    * HEAD -> Current Branch -> Latest Commit
---
* **Branch** : A **branch** in Git is a lightweight, **movable pointer to a commit**. It represents an independent line of development within the repository. Branches allow you to work on different features or experiments without affecting the main codebase. **When you create a new branch, Git creates a new pointer that points to the current commit, and you can then make changes and commits on that branch without affecting the main branch.**
    * Branch points to the latest commit in that branch, and as you make new commits, the branch pointer moves forward to point to the new commits. This allows you to **switch between branches** and work on different features or bug fixes independently.
    * So the diffrent between **HEAD** and **Branch** is that **HEAD** is a pointer to the current branch, while a **branch** is a pointer to a specific commit in the repository's history. When you switch branches, HEAD moves to point to the new branch, and the working tree and staging area are updated to reflect the state of that branch.
---
* `.git` contains `Objects\`, `References`, `HEAD`, `Configuration`, `The Index`, and more.
    * **Objects** : Contains **blobs**, **trees**, and **commits** that represent the content and history of the repository.
